```python
from bqplot import Axis, Bins, Figure, LinearScale
from ipywidgets import Layout
import numpy as np

x_scale = LinearScale()
y_scale = LinearScale()
bins = Bins(sample=[0, 1, 2, 3], bins=1, colors=['gray'],
            scales={'x': x_scale, 'y': y_scale})
fig = Figure(marks=[bins],
             axes=[Axis(scale=x_scale, label='x'),
                   Axis(scale=y_scale, label='y', orientation='vertical')],
             layout=Layout(width='400px', height='400px'))
fig
```

```python
bins.sample = np.linspace(0, 1, 20)
bins.bins = 10
```

```python
bins.bins = 4
```
